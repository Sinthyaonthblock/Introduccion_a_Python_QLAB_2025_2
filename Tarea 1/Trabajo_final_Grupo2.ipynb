**7.** Crea una nueva columna condicional `Categoria_Desarrollo`. Clasifica a los países según las siguientes reglas:
- 'Alto': Si `Esperanza_de_vida` > 80 Y `PBI_per_capita` > 25000.
- 'Medio-Alto': Si `Esperanza_de_vida` > 76 Y `PBI_per_capita` > 8000.
- 'Medio': Para todos los demás casos.

```python
'Categoria_Desarrollo'
def clasificar_desarrollo(fila):
    if fila['Esperanza_de_vida'] > 80 and fila['PBI_per_capita'] > 25000:
        return 'Alto'
    elif fila['Esperanza_de_vida'] > 76 and fila['PBI_per_capita'] > 8000:
        return 'Medio-Alto'
    else:
        return 'Medio'

df_desarrollo['Categoria_Desarrollo'] = df_desarrollo.apply(clasificar_desarrollo,axis=1)
df_desarrollo
```

**9.** Utilizando `groupby()`, calcula para cada `Continente`:

## Sección 3: Visualización de Datos

**10.** Crea un gráfico de barras horizontales (`barh`) que muestre el `Indice_Gini` de cada país. Ordena los países de menor a mayor desigualdad (Gini) para una mejor visualización. Asegúrate de que el gráfico tenga un título y etiquetas adecuadas.

```python
import matplotlib.pyplot as plt

# Ordena los países de menor a mayor desigualdad (Gini) 
df_sorted = df_desarrollo.sort_values(by="Indice_Gini", ascending=True)

# Colormap
colors = plt.cm.tab10(np.linspace(0, 1, len(df_sorted)))

# Gráfico
plt.figure(figsize=(10, 5))
bars = plt.barh(df_sorted["Pais"], df_sorted["Indice_Gini"], color=colors)
for i, (gini, pais) in enumerate(zip(df_sorted["Indice_Gini"], df_sorted["Pais"])):
    plt.text(gini + 0.5, i, f"{gini:.1f}", va='center', fontsize=10)
plt.title("Índice de Gini de cada País", fontsize=14)
plt.xlabel("Índice de Gini")
plt.ylabel("País")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**11.1.** Genera un gráfico de dispersión (scatter plot) para explorar la relación entre el `PBI_per_capita` (eje X) y la `Esperanza_de_vida` (eje Y).

**11.2.** Modifica el gráfico anterior para que el color de cada punto represente su `Continente`. (Pista: puedes crear un diccionario que mapee continentes a colores, por ejemplo: `{'Sudamérica': 'blue', ...}`).

**11.3.** **Interpretación:** ¿Qué te dice este gráfico sobre la relación entre PBI y esperanza de vida? ¿Observas alguna agrupación por continentes?

```python
# Gráfico de dispersión
plt.figure(figsize=(8,5))
plt.scatter(df_desarrollo['PBI_per_capita'],
            df_desarrollo['Esperanza_de_vida'])
for i, row in df_desarrollo.iterrows():
    plt.text(row['PBI_per_capita']+300, row['Esperanza_de_vida'], row['Pais'], fontsize=9)

# Línea de tendencia lineal
x = df_desarrollo['PBI_per_capita']
y = df_desarrollo['Esperanza_de_vida']
coef = np.polyfit(x, y, 1)
poly1d_fn = np.poly1d(coef)
x_range = np.linspace(x.min(), x.max(), 100)
plt.plot(x_range, poly1d_fn(x_range), color='red', linestyle='--', linewidth=2, label='Tendencia lineal')

# Configuración del gráfico
plt.title("El PBI per cápita y Esperanza de vida", fontsize=14)
plt.xlabel("PBI per cápita (USD)")
plt.ylabel("Esperanza de vida (años)")
```

```python
# Mapeo de continentes por colores
color_map = {
    'Sudamérica': 'blue',
    'Norteamérica': 'fuchsia',
    'Europa': 'orange',
    'Asia': 'green'
}

colors = df_desarrollo['Continente'].map(color_map)

# Gráfico de dispersión
plt.figure(figsize=(8,5))
plt.scatter(df_desarrollo['PBI_per_capita'],
            df_desarrollo['Esperanza_de_vida'],
            c=colors, s=100, alpha=0.8, edgecolor='k')

for i, row in df_desarrollo.iterrows():
    plt.text(row['PBI_per_capita']+300, row['Esperanza_de_vida'], row['Pais'], fontsize=9)

# Línea de tendencia lineal
x = df_desarrollo['PBI_per_capita']
y = df_desarrollo['Esperanza_de_vida']
coef = np.polyfit(x, y, 1)
poly1d_fn = np.poly1d(coef)
x_range = np.linspace(x.min(), x.max(), 100)
plt.plot(x_range, poly1d_fn(x_range), color='red', linestyle='--', linewidth=2, label='Tendencia lineal')

# Configuración del gráfico
plt.title("El PBI per cápita y Esperanza de vida", fontsize=14)
plt.xlabel("PBI per cápita (USD)")
plt.ylabel("Esperanza de vida (años)")

# Leyenda
for continente, color in color_map.items():
    plt.scatter([], [], c=color, label=continente, s=100, edgecolor='k')
plt.legend(title="Continente")

plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

- Los países de Europa (España) y Asia (Japón) están en la parte alta de esperanza de vida y con PBI alto.

- Los países de Sudamérica se agrupan en un rango medio-bajo tanto en PBI como en esperanza de vida.

- Norteamérica (México y Canadá) muestra un contraste: Canadá con alto PBI y alta esperanza de vida, México en un rango más bajo.

*En conclusión: la línea de tendencia lineal confirma la relación positiva: a mayor PBI per cápita, mayor esperanza de vida.

**12.** Crea un histograma de la `Tasa_Alfabetizacion`. Añade una línea vertical roja que represente la media de esta tasa.

```python
alfabetizacion = df_desarrollo['Tasa_Alfabetizacion'].dropna()

# Crear histograma
fig, ax = plt.subplots(figsize=(10, 7))
n, bins, patches = ax.hist(df_desarrollo["Tasa_Alfabetizacion"], bins=6, edgecolor="black")

# Colores distintos para cada barra
colors = plt.cm.tab20(np.linspace(0, 1, len(patches)))
for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

# Añadir etiquetas con países y tasa
for i in range(len(bins)-1):
    # Filtrar países en el intervalo
    mask = (df_desarrollo["Tasa_Alfabetizacion"] >= bins[i]) & (df_desarrollo["Tasa_Alfabetizacion"] < bins[i+1])
    paises_intervalo = df_desarrollo[mask]
    if not paises_intervalo.empty:
        texto = "\n".join([f"{p} ({t})" for p, t in zip(paises_intervalo["Pais"], paises_intervalo["Tasa_Alfabetizacion"])])
        ax.text((bins[i] + bins[i+1]) / 2, n[i] + 0.2, texto,
                ha='center', va='bottom', fontsize=8, rotation=0)

# Línea de media
media = df_desarrollo["Tasa_Alfabetizacion"].mean()
ax.axvline(media, color='red', linestyle='--', linewidth=2, label=f"Media: {media:.2f}")

# Configuración
ax.set_xlabel("Tasa de Alfabetización (%)")
ax.set_ylabel("Número de Países")
ax.set_title("Histograma de Tasa de Alfabetización por País")
ax.legend()

plt.tight_layout()
plt.show()
```
